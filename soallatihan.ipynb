```python
import numpy as np
import matplotlib.pyplot as plt

# Membaca data
data = np.loadtxt("Data/cooling.csv", delimiter=",", skiprows=1)

t = data[:, 0]
T = data[:, 1]

T_ruang = 25

# Analisis temperatur
print("Temperatur awal:", T[0], "°C")
print("Temperatur akhir:", T[-1], "°C")
print("Temperatur minimum:", np.min(T), "°C")
print("Temperatur maksimum:", np.max(T), "°C")
print("Temperatur rata-rata:", np.mean(T), "°C")

# Mencari penurunan temperatur terbesar
penurunan = T[:-1] - T[1:]
indeks = np.argmax(penurunan)

print("Penurunan terbesar:", penurunan[indeks], "°C")
print("Interval waktu:", t[indeks], "s sampai", t[indeks+1], "s")

# Menghitung logaritma temperatur
y = np.log(T - T_ruang)

# Regresi linear
kemiringan, intersep = np.polyfit(t, y, 1)

print("Kemiringan grafik:", kemiringan)

# Grafik
plt.plot(t, y, 'o', label="Data")
plt.plot(t, kemiringan*t + intersep, label="Regresi linear")

plt.xlabel("Waktu (s)")
plt.ylabel("ln(T - T_ruang)")
plt.title("Grafik Analisis Pendinginan")
plt.legend()
plt.grid()

plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

# Membaca data
data = np.loadtxt("Data/wave.csv", delimiter=",", skiprows=1)

t = data[:, 0]
y = data[:, 1]

# Menghitung amplitudo
A = (np.max(y) - np.min(y)) / 2

# Menghitung RMS
rms = np.sqrt(np.mean(y**2))

# Membandingkan RMS teoritis
rms_teori = A / np.sqrt(2)

# Menghitung perpotongan nol
titik_nol = np.where(np.diff(np.sign(y)) != 0)[0]

jumlah_potong = len(titik_nol)

# Estimasi frekuensi
durasi = t[-1] - t[0]
frekuensi = jumlah_potong / (2 * durasi)

# Menampilkan hasil
print("Amplitudo:", A)
print("RMS:", rms)
print("RMS teoritis:", rms_teori)
print("Jumlah perpotongan nol:", jumlah_potong)
print("Frekuensi:", frekuensi, "Hz")

# Grafik gelombang
plt.figure(figsize=(9, 5))
plt.plot(t, y, label="Gelombang")
plt.axhline(0, color="black", linewidth=0.8)

plt.xlabel("Waktu (s)")
plt.ylabel("Amplitudo")
plt.title("Grafik Gelombang")
plt.legend()
plt.grid()

# Simpan gambar
plt.savefig("grafik_gelombang.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Membaca data
data = np.loadtxt("Data/voltage.dat")

t = data[:, 0]
V = data[:, 1]

# Menghitung persentase penurunan tegangan
persentase = ((V[0] - V[-1]) / V[0]) * 100

print("Tegangan awal:", V[0], "V")
print("Tegangan akhir:", V[-1], "V")
print("Persentase penurunan:", persentase, "%")

# Regresi linear logaritmik
kemiringan, intersep = np.polyfit(t, np.log(V), 1)

# Menghitung konstanta waktu
tau = -1 / kemiringan

print("Kemiringan:", kemiringan)
print("Konstanta waktu tau:", tau, "s")

# Grafik skala biasa dan logaritmik
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

ax[0].plot(t, V, 'o-')
ax[0].set_title("Tegangan Skala Biasa")
ax[0].set_xlabel("Waktu (s)")
ax[0].set_ylabel("Tegangan (V)")
ax[0].grid()

ax[1].plot(t, V, 'o-')
ax[1].set_yscale("log")
ax[1].set_title("Tegangan Skala Logaritmik")
ax[1].set_xlabel("Waktu (s)")
ax[1].set_ylabel("Tegangan (V)")
ax[1].grid()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameter gerak
x0 = 2
v0 = 3
a = 1

# Membuat waktu
t = np.linspace(0, 10, 50)

# Model gerak ideal
model = x0 + v0*t + 0.5*a*t**2

# Menambahkan derau acak
rng = np.random.default_rng(1)
noise = rng.normal(0, 1, len(t))

data = model + noise

# Menghitung galat RMS
error = np.sqrt(np.mean((data - model)**2))

print("Galat RMS:", error, "m")

# Membuat grafik
plt.figure(figsize=(9, 5))

plt.scatter(t, data, label="Data berderau")
plt.plot(t, model, color="red", label="Model ideal")

plt.xlabel("Waktu (s)")
plt.ylabel("Posisi (m)")
plt.title("Model Gerak dan Data Berderau")
plt.legend()
plt.grid()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as p
# Membaca data CSV
data = np.loadtxt("electrical.csv", delimiter=",", skiprows=1)

t = data[:, 0]
V = data[:, 1]
I = data[:, 2]

# Menghitung daya listrik
P = V * I

# Analisis daya
P_max = np.max(P)
P_rata = np.mean(P)

# Menghitung energi dengan aturan trapesium
energi = np.trapezoid(P, t)

# Menampilkan hasil
print("Daya maksimum:", P_max, "Watt")
print("Daya rata-rata:", P_rata, "Watt")
print("Energi listrik:", energi, "Joule")

# Membuat grafik
plt.figure(figsize=(9, 5))

plt.plot(t, P, 'o-', label="Daya listrik")

plt.xlabel("Waktu (s)")
plt.ylabel("Daya (W)")
plt.title("Grafik Daya Listrik terhadap Waktu")
plt.legend()
plt.grid()

plt.savefig("grafik_daya.png", dpi=300, bbox_inches="tight")
plt.show()

# Menyimpan hasil ke CSV baru
hasil = np.column_stack((t, V, I, P))

np.savetxt(
    "hasil_daya.csv",
    hasil,
    delimiter=",",
    header="waktu,tegangan,arus,daya",
    comments="",
    fmt="%.6f"
)

print("Data berhasil disimpan ke hasil_daya.csv")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameter Gaussian
I0 = 100
x0 = 5
sigma = 1

# Membuat data posisi
x = np.linspace(0, 10, 1000)

# Menghitung intensitas
I = I0 * np.exp(-((x-x0)**2)/(2*sigma**2))

# Mencari posisi puncak
indeks_puncak = np.argmax(I)

x_puncak = x[indeks_puncak]
I_puncak = I[indeks_puncak]

# Menghitung intensitas relatif
intensitas_relatif = I_puncak / I0

# Mencari setengah maksimum
setengah = I_puncak / 2

# Mencari titik kiri dan kanan
kiri = np.where(I[:indeks_puncak] <= setengah)[0][-1]
kanan = indeks_puncak + np.where(
    I[indeks_puncak:] <= setengah
)[0][0]

# Interpolasi linear
x_kiri = np.interp(
    setengah,
    [I[kiri], I[kiri+1]],
    [x[kiri], x[kiri+1]]
)

x_kanan = np.interp(
    setengah,
    [I[kanan], I[kanan-1]],
    [x[kanan], x[kanan-1]]
)

# FWHM numerik
fwhm = x_kanan - x_kiri

# FWHM teoritis
fwhm_teori = 2 * np.sqrt(2*np.log(2)) * sigma

# Menampilkan hasil
print("Posisi puncak:", x_puncak)
print("Intensitas puncak:", I_puncak)
print("Intensitas relatif:", intensitas_relatif)
print("FWHM numerik:", fwhm)
print("FWHM teoritis:", fwhm_teori)

# Grafik
plt.figure(figsize=(9, 5))

plt.plot(x, I, label="Intensitas Gaussian")
plt.axhline(setengah, linestyle="--",
            label="Setengah maksimum")

plt.axvline(x_kiri, linestyle=":")
plt.axvline(x_kanan, linestyle=":")

plt.xlabel("Posisi (x)")
plt.ylabel("Intensitas")
plt.title("Grafik Intensitas Optika")
plt.legend()
plt.grid()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Menentukan parameter
x0 = 0
v0 = 2
a = 1.5

# 2. Membuat data waktu
t = np.linspace(0, 10, 50)

# 3. Menghitung posisi ideal
x_model = x0 + v0*t + 0.5*a*t**2

# 4. Membuat data dengan derau
rng = np.random.default_rng(1)
noise = rng.normal(0, 0.5, len(t))

x_data = x_model + noise

# 5. Analisis data
posisi_min = np.min(x_data)
posisi_max = np.max(x_data)
posisi_rata = np.mean(x_data)

# Menghitung kecepatan interval
kecepatan = np.diff(x_data) / np.diff(t)

# Menghitung percepatan dari model
percepatan = np.diff(kecepatan) / np.diff(t[:-1])

# Menghitung galat RMS
error = np.sqrt(np.mean((x_data-x_model)**2))

# 6. Pemeriksaan kewajaran
print("Posisi minimum:", posisi_min, "m")
print("Posisi maksimum:", posisi_max, "m")
print("Posisi rata-rata:", posisi_rata, "m")
print("Kecepatan maksimum:", np.max(kecepatan), "m/s")
print("Galat RMS:", error, "m")

if np.all(np.isfinite(x_data)) and posisi_max > posisi_min:
    print("Data valid dan memiliki rentang posisi.")
else:
    print("Data tidak valid.")

# 7. Membuat grafik
plt.figure(figsize=(9, 5))

plt.scatter(t, x_data, label="Data simulasi")
plt.plot(t, x_model, color="red", label="Model GLBB")

plt.xlabel("Waktu (s)")
plt.ylabel("Posisi (m)")
plt.title("Analisis Gerak Lurus Berubah Beraturan")
plt.legend()
plt.grid()

# Menyimpan grafik
plt.savefig("hasil_glbb.png", dpi=300, bbox_inches="tight")
plt.show()

# 8. Menyimpan hasil analisis ke CSV
hasil = np.column_stack((t, x_data, x_model))

np.savetxt(
    "hasil_glbb.csv",
    hasil,
    delimiter=",",
    header="waktu,posisi_data,posisi_model",
    comments="",
    fmt="%.6f"
)

print("Grafik berhasil disimpan.")
print("Data berhasil disimpan.")
```
